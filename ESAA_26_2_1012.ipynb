| 구분 | 구성 요소 | 설명 |
|---|---|---|
| 층 | 입력층(input layer) | 데이터를 받아들이는 층 |
| 층 | 은닉층(hidden layer) | 입력 노드로부터 입력값을 받아 가중합을 계산하고, 활성화 함수에 적용하여 출력층에 전달하는 층 |
| 층 | 출력층(output layer) | 신경망의 최종 결과값이 포함된 층 |
| 기타 | 가중치(weight) | 노드와 노드 간 연결 강도 |
| 기타 | 바이어스(bias) | 가중합에 더해 주는 상수로, 뉴런의 최종 출력값을 조절하는 역할 |
| 기타 | 가중합(weighted sum), 전달 함수 | 가중치와 신호의 곱을 합한 것 |
| 함수 | 활성화 함수(activation function) | 신호를 입력받아 적절히 처리하여 출력해 주는 함수 |
| 함수 | 손실 함수(loss function) | 출력값과 실제값 사이의 오차를 측정하는 함수 |

- 가중합 (전달 함수) : 각 노드에서 들어오는 신호에 가중치를 곱해서 다음 노드로 전달되는데 이 값들을 모두 더한 합계 -> 활성화 함수로 보냄
image.png

<div align="center">

$$ \sum_{i} w_i x_i + b $$

($w$: 가중치, $b$: 바이어스)

</div>

- 활성화 함수 : 전달 함수에서 전달받은 값을 출력할 때 일정 기준에 따라 출력 값을 변화시키는 비선형 함수 (시그모이드, 하이퍼볼릭 탄젠트, 렐루 함수 등)
  - 시그모이드 함수 : 선형 함수의 결과를 0~1 사이에서 비선형 형태로 변형
    - 로지스틱 회귀에서 주로 사용
    - 기울기 소멸 문제가 있음
  <div align="center">

  $$ f(x) = \frac{1}{1 + e^{-x}} $$

  </div>

    - 하이퍼볼릭 탄젠트 함수 : 선형 함수의 결과를 -1~1 사이에서 비선형 형태로 변형
      - 양수 편향 문제는 해결되었지만 기울기 소멸 문제는 여전히 발생
    - 렐루 함수 : 입력(x)이 음수일 때는 0을 출력하고 양수일 때는 x를 출력
      - 학습 속도가 빠르고 기울기 소멸 문제 발생 X
      - 일반적으로 은닉층에서 사용되며 하이퍼볼릭 탄젠트 함수 대비 학습 속도가 6배 빠름
      - 문제점 : 음수 값을 입력받으면 항상 0을 출력 받아 학습 능력이 감소
    - 리키 렐루 함수 : 입력 값이 음수이면 0이 아닌 0.001처럼 매우 작은 수를 반환
      - 입력 값이 수렴하는 구간이 제거되어 렐루 함수를 사용할 때 생기는 문제 해결
    - 소프트맥스 함수 : 입력 값을 0~1 사이에 출력되도록 정규화하여 출력 값들의 총합이 항상 1이 되도록 함
      - 출력 노드의 활성화 함수로 많이 사용됨
<div align="center">

$$ y_k = \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)} $$

</div>

```python
import torch
```

```python
#렐루 함수 / 소프트 맥스 함수 구현
class Net(torch.nn.Module):
  def __init__(self, n_feature, n_hidden, n_output):
    super(Net, self).__init__()
    self.hidden = torch.nn.Linear(n_feature, n_hidden)
    self.relu = torch.nn.ReLU(inplace=True)
    self.out = torch.nn.Linear(n_hidden, n_output)
    self.softmax = torch.nn.Softmax(dim=n_output)
  def forward(self, x):
    x = self.hidden(x)
    x = self.relu(x)
    x = self.out(x)
    x = self.softmax(x)
    return x
```

- 머신 러닝에서는 비선형 분류를 하기 위해 여러 트릭을 사용했지만 딥러닝은 다수의 은닉층을 추가했기 때문에 별도의 트릭 없이 비선형 분류가 가능

**합성곱 신경망**
- 합성곱층과 풀링층을 포함하는 이미지 처리 성능이 좋은 인공 신경망 알고리즘
- 영상 및 사진이 포함된 이미지 데이터에서 객체를 탐색하거나 객체 위치를 찾아내는 데 유용

**순환 신경망**

- 순환 : 현재 결과가 이전 결과와 연관이 있다는 의미
- 문제점 : 기울기 소멸 문제
- ex) 언어 모델링, 텍스트 생성, 자동 번역, 음성 인식, 이미지 캡션 생성 등
