# **합성곱 신경망I**

### (1) 합성곱 신경망

- 순전파 과정의 복잡한 계산 과정과 많은 요구 자원, 오랜 계산 시간을 해결하기 위함
- 이미지 전체를 한 번에 계산하는 것이 아닌 이미지의 국소적 부분을 계산함으로써 시간과 자원을 절약하며 이미지의 세밀한 부분까지 분석할 수 있는 신경망

**합성곱층의 필요성**

- 합성곱 신경망 : 이미지나 영상을 처리하는 데 유용
- 이미지 분석 : 배열을 펼쳐서 각 픽셀에 가중치를 곱하여 은닉층으로 전달 -> 이때 이미지를 펼침으로써 발생하는 데이터의 공간적 구조 무시를 방지하기 위해 합성곱층 도입

**합성곱 신경망 구조**

- 합성곱 신경망 : 음성 인식이나 이미지/영상 인식에서 주로 사용되는 신경망
- 계층 : 입력층, 합성곱층, 풀링층, 완전연결층, 출력층

[합성곱층]
- 입력 데이터에서 특성을 추출하는 역할을 수행
- 특성 추출 : 입력 이미지가 들어왔을 때 이미지에 대한 특성을 감지하기 위해 커널이나 필터 사용
- 특성 맵 : 커널/필터가 이미지의 모든 영역을 훑으면서 추출한 결과물

[스트라이드가 1일 때 이동하는 과정]
1. 입력 이미지에 3X3 필터 적용
- 입력 이미지와 필터를 포개 놓고 대응되는 숫자끼리 곱한 후 모두 더함
2. 필터가 1만큼 이동
3. 필터가 1만큼 두 번째 이동
4. 필터가 1만큼 세 번째 이동
5. 필터가 1만큼 네 번째 이동
6. 필터가 1만큼 마지막으로 이동

[풀링층]
- 합성곱층과 유사하게 특성 맵의 차원을 다운 샘플링하여 연산량 감소
- 최대 풀링: 대상 영역에서 최댓값을 추출
- 평균 풀링: 대상 영역에서 평균을 반환

image.png

[완전연결층]
- 이미지는 3차원 벡터에서 1차원 벡터로 펼쳐짐

[출력층]
- 소프트맥스 활성화 함수를 사용해 입력받은 값을 0~1 사이의 값으로 출력
- 마지막 출력층의 소프트맥스 함수를 사용하여 이미지가 각 레이블에 속할 확률 값이 출력되며 이때 가장 높은 확률 값을 갖는 레이블이 최종 값으로 선정

**1D, 2D, 3D 합성곱**

[1D 합성곱]
- 필터가 시간을 축으로 좌우로만 이동할 수 있는 합성곱
- 출력 W : 입력 W와 필터 K에 대한 출력

[2D 합성곱]
- 필터가 방향 두 개로 움직이는 형태
- 출력 (W, H) : 입력(W, H)과 필터(k, k)에 대한 출력

[3D 합성곱]
- 필터가 움직이는 방향이 세 개
- 출력(W, H, L) : 입력(W, H, L)에 대해 필터(k, k, d)를 적용

[3D 입력을 갖는 2D 합성곱]
- 입력이 3D 형태임에도 출력 형태가 3D가 아닌 2D 행렬을 취하는 것

[1X1 합성곱]
- 3D 형태로 입력
- 출력(W, H) : 입력(W, H, L)에 필터(1, 1, L)를 적용
